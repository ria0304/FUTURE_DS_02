```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots

np.random.seed(42)

def generate_social_media_data(n_campaigns=200):
    platforms = ['Instagram', 'Facebook', 'Twitter', 'LinkedIn']
    campaign_types = ['Brand Awareness', 'Conversion', 'Engagement', 'Traffic']
    ad_types = ['Image', 'Video', 'Carousel', 'Story']
    
    data = []
    
    for i in range(n_campaigns):
        platform = np.random.choice(platforms)
        campaign_type = np.random.choice(campaign_types)
        ad_type = np.random.choice(ad_types)
        
        impressions = np.random.randint(1000, 500000)
        reach = int(impressions * np.random.uniform(0.3, 0.8))
        clicks = np.random.randint(50, 50000)
        engagements = np.random.randint(100, 100000)
        spend = np.random.uniform(50, 5000)
        conversions = np.random.randint(0, 1000)
        
        ctr = (clicks / impressions) * 100 if impressions > 0 else 0
        cpc = spend / clicks if clicks > 0 else 0
        cpe = spend / engagements if engagements > 0 else 0
        conversion_rate = (conversions / clicks) * 100 if clicks > 0 else 0
        roas = (conversions * 50) / spend if spend > 0 else 0
        
        data.append({
            'campaign_id': f'CMP_{i+1:03d}',
            'platform': platform,
            'campaign_type': campaign_type,
            'ad_type': ad_type,
            'impressions': impressions,
            'reach': reach,
            'clicks': clicks,
            'engagements': engagements,
            'spend': round(spend, 2),
            'conversions': conversions,
            'ctr': round(ctr, 2),
            'cpc': round(cpc, 2),
            'cpe': round(cpe, 2),
            'conversion_rate': round(conversion_rate, 2),
            'roas': round(roas, 2)
        })
    
    return pd.DataFrame(data)

campaign_df = generate_social_media_data()

class SocialMediaAnalyzer:
    def __init__(self, df):
        self.df = df
    
    def performance_summary(self):
        total_spend = self.df['spend'].sum()
        total_clicks = self.df['clicks'].sum()
        total_conversions = self.df['conversions'].sum()
        total_impressions = self.df['impressions'].sum()
        
        avg_ctr = (total_clicks / total_impressions) * 100
        avg_cpc = total_spend / total_clicks
        overall_roas = (total_conversions * 50) / total_spend
        
        summary = {
            'Total Spend': f'${total_spend:,.2f}',
            'Total Clicks': f'{total_clicks:,}',
            'Total Conversions': f'{total_conversions:,}',
            'Total Impressions': f'{total_impressions:,}',
            'Average CTR': f'{avg_ctr:.2f}%',
            'Average CPC': f'${avg_cpc:.2f}',
            'Overall ROAS': f'{overall_roas:.2f}x'
        }
        
        return summary
    
    def platform_comparison(self):
        platform_stats = self.df.groupby('platform').agg({
            'spend': 'sum',
            'clicks': 'sum',
            'conversions': 'sum',
            'impressions': 'sum',
            'ctr': 'mean',
            'roas': 'mean'
        }).round(2)
        
        platform_stats['cpc'] = platform_stats['spend'] / platform_stats['clicks']
        return platform_stats

def create_social_media_dashboard(df):
    fig = make_subplots(
        rows=3, cols=3,
        subplot_titles=(
            'Spend Distribution by Platform', 'CTR by Campaign Type',
            'ROAS by Platform', 'Conversions vs Spend',
            'CPC by Ad Type', 'Campaign Type Distribution',
            'Engagement Rate by Platform', 'Conversion Rate vs Spend', 'Performance Heatmap'
        ),
        specs=[
            [{"type": "pie"}, {"type": "bar"}, {"type": "bar"}],
            [{"type": "scatter"}, {"type": "bar"}, {"type": "pie"}],
            [{"type": "bar"}, {"type": "scatter"}, {"type": "heatmap"}]
        ]
    )
    
    platform_spend = df.groupby('platform')['spend'].sum()
    fig.add_trace(
        go.Pie(labels=platform_spend.index, values=platform_spend.values),
        row=1, col=1
    )
    
    ctr_by_type = df.groupby('campaign_type')['ctr'].mean()
    fig.add_trace(
        go.Bar(x=ctr_by_type.index, y=ctr_by_type.values),
        row=1, col=2
    )
    
    roas_by_platform = df.groupby('platform')['roas'].mean()
    fig.add_trace(
        go.Bar(x=roas_by_platform.index, y=roas_by_platform.values),
        row=1, col=3
    )
    
    fig.add_trace(
        go.Scatter(x=df['spend'], y=df['conversions'], mode='markers',
                  marker=dict(size=8, color=df['ctr'], colorscale='Viridis', showscale=True)),
        row=2, col=1
    )
    
    cpc_by_ad_type = df.groupby('ad_type')['cpc'].mean()
    fig.add_trace(
        go.Bar(x=cpc_by_ad_type.index, y=cpc_by_ad_type.values),
        row=2, col=2
    )
    
    type_distribution = df['campaign_type'].value_counts()
    fig.add_trace(
        go.Pie(labels=type_distribution.index, values=type_distribution.values),
        row=2, col=3
    )
    
    df['engagement_rate'] = (df['engagements'] / df['reach']) * 100
    engagement_by_platform = df.groupby('platform')['engagement_rate'].mean()
    fig.add_trace(
        go.Bar(x=engagement_by_platform.index, y=engagement_by_platform.values),
        row=3, col=1
    )
    
    fig.add_trace(
        go.Scatter(x=df['spend'], y=df['conversion_rate'], mode='markers',
                  marker=dict(size=8, color=df['roas'], colorscale='Plasma', showscale=True)),
        row=3, col=2
    )
    
    heatmap_data = df.pivot_table(values='roas', index='platform', columns='campaign_type', aggfunc='mean')
    fig.add_trace(
        go.Heatmap(z=heatmap_data.values, x=heatmap_data.columns, y=heatmap_data.index,
                  colorscale='RdYlGn'),
        row=3, col=3
    )
    
    fig.update_layout(height=1200, title_text="Social Media Campaign Performance Dashboard", showlegend=False)
    return fig

analyzer = SocialMediaAnalyzer(campaign_df)

print("PERFORMANCE SUMMARY")
print("=" * 50)
summary = analyzer.performance_summary()
for key, value in summary.items():
    print(f"{key}: {value}")

print("\nPLATFORM PERFORMANCE")
print("=" * 50)
platform_stats = analyzer.platform_comparison()
print(platform_stats)

print("\nTOP 5 CAMPAIGNS BY ROAS")
print("=" * 50)
top_campaigns = campaign_df.nlargest(5, 'roas')[['campaign_id', 'platform', 'campaign_type', 'roas', 'spend', 'conversions']]
print(top_campaigns.to_string(index=False))

dashboard = create_social_media_dashboard(campaign_df)
dashboard.show()

def calculate_kpis(df):
    kpis = {}
    
    kpis['Overall CTR'] = f"{(df['clicks'].sum() / df['impressions'].sum() * 100):.2f}%"
    kpis['Overall Conversion Rate'] = f"{(df['conversions'].sum() / df['clicks'].sum() * 100):.2f}%"
    kpis['Average CPC'] = f"${(df['spend'].sum() / df['clicks'].sum()):.2f}"
    kpis['Average ROAS'] = f"{(df['roas'].mean()):.2f}x"
    kpis['Engagement Rate'] = f"{(df['engagements'].sum() / df['reach'].sum() * 100):.2f}%"
    kpis['Cost per Engagement'] = f"${(df['spend'].sum() / df['engagements'].sum()):.2f}"
    
    return kpis

def performance_insights(df):
    insights = []
    
    best_platform_roas = df.groupby('platform')['roas'].mean().idxmax()
    worst_platform_roas = df.groupby('platform')['roas'].mean().idxmin()
    insights.append(f"Best performing platform by ROAS: {best_platform_roas}")
    insights.append(f"Platform needing optimization: {worst_platform_roas}")
    
    best_type_ctr = df.groupby('campaign_type')['ctr'].mean().idxmax()
    insights.append(f"Highest CTR campaign type: {best_type_ctr}")
    
    most_efficient_cpc = df.groupby('platform')['cpc'].mean().idxmin()
    insights.append(f"Most cost-efficient platform: {most_efficient_cpc}")
    
    best_ad_type_engagement = df.groupby('ad_type')['engagement_rate'].mean().idxmax()
    insights.append(f"Best performing ad type: {best_ad_type_engagement}")
    
    return insights

print("\nKEY PERFORMANCE INDICATORS")
print("=" * 50)
kpis = calculate_kpis(campaign_df)
for kpi, value in kpis.items():
    print(f"{kpi}: {value}")

print("\nACTIONABLE INSIGHTS")
print("=" * 50)
insights = performance_insights(campaign_df)
for insight in insights:
    print(f"- {insight}")

print("\nRECOMMENDATIONS")
print("=" * 50)
print("1. Reallocate budget towards higher ROAS platforms")
print("2. Optimize underperforming campaign types")
print("3. Focus on improving CTR through better ad creatives")
print("4. Implement A/B testing for low-performing platforms")
print("5. Analyze top performing campaigns for best practices")

campaign_df.to_csv('social_media_campaign_data.csv', index=False)
print("\nData exported to social_media_campaign_data.csv")
```
